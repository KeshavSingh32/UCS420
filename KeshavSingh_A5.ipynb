```python
import numpy as np
```

```python
#Q1
arr = np.array([10, 20, 30, 40, 50])

print("Original Array:")
print(arr)

print("\nAddition of 2:")
print(arr + 2)

print("\nMultiplication by 3:")
print(arr * 3)

print("\nDivision by 2:")
print(arr / 2)
```

```python
#Q2 (a)
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:")
print(arr)

print("Reversed Array:")
print(arr[::-1])

#(b)->(i)
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print("Most Frequent Value:", most_frequent)

print("Indices:", indices)

#(b)->(ii)
import numpy as np

y = np.array([1,1,1,2,3,4,2,4,3,3])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(y == most_frequent)[0]

print("Most Frequent Value:", most_frequent)

print("Indices:", indices)


```

```python
#Q3
arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print("Array:")
print(arr)

print("\n1st row, 2nd column:")
print(arr[0,1])

print("\n3rd row, 1st column:")
print(arr[2,0])
```

```python
#Q4


Keshav = np.linspace(10,100,25)

print("Array:")
print(Keshav)

print("\nDimensions:", Keshav.ndim)

print("Shape:", Keshav.shape)

print("Total Elements:", Keshav.size)

print("Data Type:", Keshav.dtype)

print("Total Bytes:", Keshav.nbytes)

transpose = Keshav.reshape(25,1)

print("\nTranspose using reshape():")
print(transpose)

print("\nUsing T attribute:")
print(Keshav.T)
```

```python
#Q5
ucs420_Keshav = np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])

print("Original Array:")
print(ucs420_Keshav)

print("\nMean:", np.mean(ucs420_Keshav))

print("Median:", np.median(ucs420_Keshav))

print("Maximum:", np.max(ucs420_Keshav))

print("Minimum:", np.min(ucs420_Keshav))

print("Unique Elements:")
print(np.unique(ucs420_Keshav))

reshaped_ucs420_Keshav = ucs420_Keshav.reshape(4,3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_Keshav)

resized_ucs420_Keshav = np.resize(ucs420_Keshav, (2,3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_Keshav)
```
