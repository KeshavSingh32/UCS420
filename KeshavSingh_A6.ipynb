```python
import numpy as np
import pandas as pd
```

```python
#Q1

import matplotlib.pyplot as plt

M = int(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(8,5))

plt.plot(x, y1, color="blue", linestyle="-", label="M*x^2")
plt.plot(x, y2, color="red", linestyle="--", label="M*sin(x)")

plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Q2
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Subject": ["Math", "Physics", "Chemistry", "English", "Computer"],
    "Score": [88, 75, 92, 81, 95]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8,5))

ax = sns.barplot(x="Subject", y="Score", data=df, hue="Subject", palette="Set2", legend=False)

for i, value in enumerate(df["Score"]):
    ax.text(i, value + 1, str(value), ha="center")

plt.title("Subject Scores")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(True)

plt.show()
```

```python
#Q3
np.random.seed(1024170166)

data = np.random.randn(50)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(np.cumsum(data))
plt.title("Cumulative Sum")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)

plt.subplot(2,2,2)
plt.scatter(range(50), data)
plt.title("Scatter Plot")
plt.xlabel("Index")
plt.ylabel("Random Values")
plt.grid(True)

plt.subplot(2,2,3)
plt.hist(data)
plt.title("Histogram")
plt.grid(True)

plt.subplot(2,2,4)
plt.boxplot(data)
plt.title("Box Plot")
plt.grid(True)

plt.tight_layout()

plt.show()
```
