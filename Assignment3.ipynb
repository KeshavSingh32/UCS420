```python
import pandas as pd

# Q.6
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25",
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5],
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)


df = pd.read_csv("employees.csv")

# (a)
print("a) Shape:", df.shape)

# (b
print("\nb) Summary:")
df.info()

# (c)
print("\nc) Descriptive Statistics:")
print(df.describe())

# (d)
print("\nd) First 5 rows:\n", df.head(5))
print("\nd) Last 3 rows:\n", df.tail(3))

# (e)
print("\ne) i. Average Salary:", df["Salary"].mean())
print("e) ii. Total Bonus:", df["Bonus"].sum())
print("e) iii. Youngest Employee Age:", df["Age"].min())
print("e) iv. Highest Performance Rating:", df["Rating"].max())

# (f)
print(
    "\nf) Sorted by Salary (Descending):\n",
    df.sort_values(by="Salary", ascending=False),
)


# (g)
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


df["Performance_Category"] = df["Rating"].apply(categorize_rating)
print(
    "\ng) Performance Categories:\n", df[["Name", "Rating", "Performance_Category"]]
)

# (h)
print("\nh) Missing Values Count:\n", df.isnull().sum())

# (i)
df = df.rename(columns={"Employee_ID": "ID"})

# (j)
print("\nj) i. Experience > 5 years:\n", df[df["Years_of_Experience"] > 5])
print("\nj) ii. IT Department employees:\n", df[df["Department"] == "IT"])

# (k)
df["Tax"] = df["Salary"] * 0.10

# (l)
df.to_csv("modified_employees.csv", index=False)
print("\nl) Modified DataFrame saved to 'modified_employees.csv'")
```
