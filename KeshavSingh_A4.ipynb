```python
import pandas as pd
```

```python
#Q1

fixed_enteries=[
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing",
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account",
    },

    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general",
    },

    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing",
    },

]

df=pd.DataFrame(fixed_enteries)

roll_number = "1024170166"
categories=["billing","account","general"]
custom_enteries=[]

for d in map(int,roll_number[-2:]):

  category=categories[d%3]

  if category == "billing":
        question = "How can I download my fee receipt?"
        answer = "Log in to the student portal and click 'Download Receipt'."
        keywords = "receipt fee download"

  elif category == "account":
        question = "How do I update my registered mobile number?"
        answer = "Go to Account Settings and update your mobile number."
        keywords = "mobile update account"

  else:
        question = "Where is the administration office located?"
        answer = "The administration office is in Block A."
        keywords = "office location admin"

  custom_enteries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

cus_df=pd.DataFrame(custom_enteries)

final_df=pd.concat([df,cus_df],ignore_index=True)
print(final_df)

```

```python
#Q2

def score_query(query, df):
    query_words = query.lower().split()

    results = []

    for _, row in df.iterrows():

        score = 0

        text = (
            row["question"].lower() + " " +
            row["keywords"].lower() + " " +
            row["answer"].lower()
        )

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": score
            })

    results = sorted(results, key=lambda x: x["confidence"], reverse=True)

    return pd.DataFrame(results)


query = input("Enter your query: ")

print(score_query(query, final_df))
```

```python
# Q3
def same_category(category_name, df):

    return df[df["category"] == category_name]


category = custom_enteries[0]["category"]

print(same_category(category, final_df))
```

```python
# Q4
print(final_df)

entry = int(input("Enter row number to update: "))

new_keyword = input("Enter new keyword: ")

final_df.loc[entry, "keywords"] = (
    final_df.loc[entry, "keywords"] + " " + new_keyword
)

filename = roll_number + "_faq_data.csv"

final_df.to_csv(filename, index=False)

print("Updated DataFrame")

print(final_df)

print("Saved as", filename)
```

```python
# Q6
def score_query_with_ties(query, df):

    query_words = query.lower().split()

    results = []

    for _, row in df.iterrows():

        score = 0

        text = (
            row["question"].lower() + " " +
            row["keywords"].lower() + " " +
            row["answer"].lower()
        )

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": score
            })

    if len(results) == 0:
        print("No matching entries found.")
        return

    result_df = pd.DataFrame(results)

    highest = result_df["confidence"].max()

    best = result_df[result_df["confidence"] == highest]

    print(best)


print("Tie Example")

score_query_with_ties("fee", final_df)

print()

print("Non-Tie Example")

score_query_with_ties("password", final_df)
```
